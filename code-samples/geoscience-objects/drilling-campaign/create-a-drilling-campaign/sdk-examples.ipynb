### Define helper functions

These functions assist with assembling the elements and components of geoscience objects and for viewing the new object in the Evo portal.

```python
import numpy as np
import pandas as pd


def create_hole_id_mapping(hole_id_table, value_list):
    """
    Create a hole ID mapping table based on the hole ID table and the value list.

    Args:
        hole_id_table (pd.DataFrame): The hole ID lookup table.
        value_list (pd.DataFrame): The value list to create the mapping from.

    Returns:
        mapping_df (pd.DataFrame): The hole ID mapping table.
    """

    num_keys = len(hole_id_table.index)

    mapping_df = pd.DataFrame(list())
    mapping_df["hole_index"] = hole_id_table["key"]
    mapping_df["offset"] = [0] * num_keys
    mapping_df["count"] = [0] * num_keys

    mapping_df["hole_index"] = mapping_df["hole_index"].astype("int32")
    mapping_df["offset"] = mapping_df["offset"].astype("uint64")
    mapping_df["count"] = mapping_df["count"].astype("uint64")

    prev_value = ""
    key = ""
    count = 0
    offset = 0

    for index, row in value_list.iterrows():
        new_value = row["data"]

        if new_value != prev_value:
            if prev_value != "":
                mapping_df.loc[mapping_df["hole_index"] == key, "count"] = count
                mapping_df.loc[mapping_df["hole_index"] == key, "offset"] = offset
                offset += count

            mask = hole_id_table["value"] == new_value
            masked_df = hole_id_table[mask]
            try:
                key_row = masked_df.iloc[[0]]
            except IndexError:
                print("Ignoring this hole ID")
                continue

            key = key_row["key"].iloc[0]
            count = 1
            prev_value = new_value
        else:
            count += 1

    mapping_df.loc[mapping_df["hole_index"] == key, "count"] = count
    mapping_df.loc[mapping_df["hole_index"] == key, "offset"] = offset

    return mapping_df


def create_category_lookup_and_values(attribute):
    """
    Create a category lookup table and the associated column of mapped key values.

    Args:
        attribute (pd.DataFrame): An attribute of a geoscience object.

    Returns:
        table_df (pd.DataFrame): The category lookup table.
        values_df (pd.DataFrame): The associated column with mapped key values.
    """

    # Replace NaN with empty string
    attribute.replace(np.nan, "", regex=True, inplace=True)
    set_obj = set(attribute["data"])
    list_obj = list(set_obj)
    list_obj.sort()
    num_unique_elements = len(list_obj)

    # Create lookup table
    table_df = pd.DataFrame([])
    table_df["key"] = list(range(0, num_unique_elements))
    table_df["value"] = list_obj

    # Create data column
    values_df = pd.DataFrame([])
    values_df["data"] = attribute["data"].map(table_df.set_index("value")["key"])
    return table_df, values_df


def create_hole_chunks_df(collar_hole_ids, path_hole_ids, hole_id_table):
    """
    Build a hole-chunks dataframe with one row per collar hole.

    Each chunk records, for a single drillhole, its index into the campaign hole ID lookup table
    and the range of rows it occupies in the path (segment) table.

    Args:
        collar_hole_ids (Iterable): The hole ID of each collar, in collar/output order.
        path_hole_ids (Iterable): The hole ID of each path row, in stored path order. Rows that
            belong to the same hole are assumed to be contiguous (sort the path table by hole ID).
        hole_id_table (pd.DataFrame): The union hole ID lookup table (columns "key" and "value").

    Returns:
        chunks_df (pd.DataFrame): Columns hole_index (int32), offset (uint64), count (uint64).
    """

    value_to_key = dict(zip(hole_id_table["value"], hole_id_table["key"]))

    offsets = {}
    counts = {}
    for row_index, hole_id in enumerate(path_hole_ids):
        if hole_id not in counts:
            counts[hole_id] = 0
            offsets[hole_id] = row_index
        counts[hole_id] += 1

    chunks_df = pd.DataFrame()
    chunks_df["hole_index"] = [value_to_key[hole_id] for hole_id in collar_hole_ids]
    chunks_df["offset"] = [offsets.get(hole_id, 0) for hole_id in collar_hole_ids]
    chunks_df["count"] = [counts.get(hole_id, 0) for hole_id in collar_hole_ids]

    chunks_df["hole_index"] = chunks_df["hole_index"].astype("int32")
    chunks_df["offset"] = chunks_df["offset"].astype("uint64")
    chunks_df["count"] = chunks_df["count"].astype("uint64")
    return chunks_df


def create_attributes(dataframe, reserved_columns, data_client):
    """
    Publish every column of ``dataframe`` that is not a reserved column as an attribute.

    The attribute type is detected automatically from each column:

    - Numeric columns are published as continuous attributes.
    - Text columns are published as category attributes when they have fewer than half as many
      unique values as rows, and as plain string attributes otherwise. This keeps low-cardinality
      text (a handful of repeated labels) compact via a lookup table, while high-cardinality text
      (mostly unique values) avoids a lookup table that is almost as large as the data itself.

    Each attribute has one value per row in ``dataframe``, so collar attributes match the number of
    holes and survey attributes match the number of surveys.

    Args:
        dataframe (pd.DataFrame): The source table (one row per collar or per survey).
        reserved_columns (Iterable[str]): Column names that are handled explicitly and must not be
            published as attributes.
        data_client: The data client used to persist attribute data.

    Returns:
        attributes (list): The attribute components, one per extra column.
    """

    from evo_schemas.components import (
        CategoryAttribute_V1_1_0,
        ContinuousAttribute_V1_1_0,
        NanCategorical_V1_0_1,
        NanContinuous_V1_0_1,
        StringAttribute_V1_1_0,
    )
    from evo_schemas.elements import (
        FloatArray1_V1_0_1,
        IntegerArray1_V1_0_1,
        LookupTable_V1_0_1,
        StringArray_V1_0_1,
    )

    reserved = set(reserved_columns)
    attributes = []

    for name in dataframe.columns:
        if name in reserved:
            continue

        key = str(uuid.uuid4())
        column = dataframe[name]

        if pd.api.types.is_numeric_dtype(column):
            values_df = pd.DataFrame()
            values_df["data"] = pd.to_numeric(column, errors="coerce").astype("float64")
            values = FloatArray1_V1_0_1.from_dict(data_client.save_dataframe(values_df))
            attributes.append(
                ContinuousAttribute_V1_1_0(
                    name=name, key=key, nan_description=NanContinuous_V1_0_1(values=[]), values=values
                )
            )
        else:
            attribute_df = pd.DataFrame()
            attribute_df["data"] = column
            attribute_df = attribute_df.astype(str).reset_index(drop=True)

            num_rows = len(attribute_df.index)
            num_unique = attribute_df["data"].nunique()

            # Low-cardinality text becomes a category attribute (lookup table plus keys). Text with
            # at least half as many unique values as rows becomes a plain string attribute instead.
            if num_unique < num_rows / 2:
                table_df, values_df = create_category_lookup_and_values(attribute_df)

                table = LookupTable_V1_0_1.from_dict(data_client.save_dataframe(table_df))
                values = IntegerArray1_V1_0_1.from_dict(data_client.save_dataframe(values_df))
                attributes.append(
                    CategoryAttribute_V1_1_0(
                        name=name,
                        key=key,
                        nan_description=NanCategorical_V1_0_1(values=[]),
                        table=table,
                        values=values,
                    )
                )
            else:
                values = StringArray_V1_0_1.from_dict(data_client.save_dataframe(attribute_df))
                attributes.append(StringAttribute_V1_1_0(name=name, key=key, values=values))

    return attributes
```

### Attributes

```python
import pandas as pd
from evo_schemas.components import CategoryAttribute_V1_1_0, NanCategorical_V1_0_1, StringAttribute_V1_1_0
from evo_schemas.elements import IntegerArray1_V1_0_1, LookupTable_V1_0_1, StringArray_V1_0_1

attributes = []

for collar_attribute in collar_attributes:
    name = collar_attribute["name"]
    key = collar_attribute["key"]
    type = collar_attribute["type"]

    attribute_df = pd.DataFrame()
    attribute_df["data"] = input_df[name]

    if type == "string":
        attribute_df = attribute_df.astype(str).reset_index(drop=True)
        values = StringArray_V1_0_1.from_dict(data_client.save_dataframe(attribute_df))
        attribute = StringAttribute_V1_1_0(name=name, key=key, values=values)
    elif type == "category":
        attribute_df = attribute_df.astype(str).reset_index(drop=True)
        table_df, values_df = create_category_lookup_and_values(attribute_df)

        table = LookupTable_V1_0_1.from_dict(data_client.save_dataframe(table_df))
        values = IntegerArray1_V1_0_1.from_dict(data_client.save_dataframe(values_df))

        attribute = CategoryAttribute_V1_1_0(
            name=name, key=key, nan_description=NanCategorical_V1_0_1(values=[]), table=table, values=values
        )
    else:
        print(f"Unsupported attribute type: {type}")
        continue

    attributes.append(attribute)
```

## Interim data (optional)

Once drilling is under way, interim data captures where the holes have actually been drilled. It is provided as two files:

- **Interim collar file** with the columns `hole_id`, `x`, `y`, `z`, `final`, `target`, `current` (one row per interim hole).
- **Interim survey file** with the columns `hole_id`, `depth`, `dip`, `azimuth` (one row per survey).

This data is optional. If the interim files are present in `input_path`, the interim section is built and the campaign `hole_id` component is rebuilt as the union of the planned and interim hole IDs. If they are absent, only the planned section is published. Either way, the object is assembled and published in a single step at the end of the notebook.

Any columns in the interim files that are not part of the fixed schema are automatically published as attributes. Interim collar attributes have one value per interim hole, and interim survey attributes have one value per survey.


```python
from IPython.display import display

# Interim data is optional. The interim collar file defines the interim drillholes with the fixed
# columns hole_id, x, y, z, final, target and current (one row per hole). Any other columns are
# published as collar attributes. We only build the interim section when the interim files exist.
interim_collar_hole_id_column = "hole_id"
interim_survey_hole_id_column = "hole_id"
interim_collar_file_path = f"{input_path}/interim-collar.csv"
interim_survey_file_path = f"{input_path}/interim-survey.csv"

# Defaults so the single publish at the end works even when there is no interim data.
has_interim = os.path.exists(interim_collar_file_path) and os.path.exists(interim_survey_file_path)
object_hole_id_component = hole_id_component  # Planned-only hole IDs unless interim data is added.
interim_component = None


def adapt_leapfrog_collar(collar_df):
    """Convert a Leapfrog-exported collar dataframe (holeid, x, y, z, maxdepth) to the fixed schema."""
    collar_df = collar_df.rename(columns={"holeid": interim_collar_hole_id_column})
    collar_df["final"] = collar_df["maxdepth"]
    collar_df["target"] = collar_df["maxdepth"]
    collar_df["current"] = collar_df["maxdepth"]
    return collar_df.drop(columns=["maxdepth"])


def adapt_leapfrog_survey(survey_df):
    """Convert a Leapfrog-exported survey dataframe (holeid, depth, dip, azimuth) to the fixed schema."""
    return survey_df.rename(columns={"holeid": interim_survey_hole_id_column})


# When the collar file is in the Leapfrog format (a maxdepth column), the matching survey file uses a
# holeid column too, so the same adapter is applied to both files.
use_leapfrog_adapter = False

if has_interim:
    interim_collar_df = pd.read_csv(interim_collar_file_path)

    # Leapfrog exports interim collar files with the columns holeid, x, y, z, maxdepth instead of the
    # fixed schema above. Detect this format by the presence of a maxdepth column and adapt it: rename
    # holeid to hole_id and use the single maxdepth value for the final, target and current columns.
    use_leapfrog_adapter = "maxdepth" in interim_collar_df.columns
    if use_leapfrog_adapter:
        interim_collar_df = adapt_leapfrog_collar(interim_collar_df)

    interim_collar_df = interim_collar_df.sort_values([interim_collar_hole_id_column]).reset_index(drop=True)

    num_interim_holes = len(interim_collar_df.index)
    display(interim_collar_df.head())
else:
    print(f"No interim files found in '{input_path}'. The planned section will be published on its own.")
```

```python
# The interim survey file contains the downhole survey measurements. It has one row per survey and
# the fixed columns hole_id, depth, dip and azimuth. Any other columns are published as survey
# attributes.
if has_interim:
    interim_survey_df = pd.read_csv(interim_survey_file_path)

    # Apply the Leapfrog adapter to the survey file too when the collar file was in the Leapfrog
    # format, so its holeid column is renamed to hole_id.
    if use_leapfrog_adapter:
        interim_survey_df = adapt_leapfrog_survey(interim_survey_df)

    # Keep only surveys for holes that are defined in the interim collar file, then sort by hole ID so
    # that each hole's surveys are stored contiguously.
    interim_survey_df = interim_survey_df[
        interim_survey_df[interim_survey_hole_id_column].isin(interim_collar_df[interim_collar_hole_id_column])
    ]

    interim_survey_df = interim_survey_df.sort_values([interim_survey_hole_id_column]).reset_index(drop=True)

    display(interim_survey_df.head())
    num_interim_surveys = len(interim_survey_df.index)
```

### Combine planned and interim hole IDs

The campaign `hole_id` component must list every hole across the whole campaign. Because interim holes are not required to match the planned holes, we rebuild the lookup table as the union of the planned and interim hole IDs. The planned holes keep the keys assigned earlier, and any interim-only holes are appended with new keys.


```python
from evo_schemas.components import CategoryData_V1_0_1
from evo_schemas.elements import IntegerArray1_V1_0_1, LookupTable_V1_0_1

if has_interim:
    # Append any interim-only hole IDs to the planned hole ID lookup table.
    planned_hole_ids = set(hole_id_table_df["value"])
    interim_only_ids = [
        hole_id for hole_id in interim_collar_df[interim_collar_hole_id_column] if hole_id not in planned_hole_ids
    ]

    union_hole_id_table_df = pd.concat(
        [
            hole_id_table_df,
            pd.DataFrame(
                {
                    "key": range(len(hole_id_table_df), len(hole_id_table_df) + len(interim_only_ids)),
                    "value": interim_only_ids,
                }
            ),
        ],
        ignore_index=True,
    )
    union_hole_id_table_df["key"] = union_hole_id_table_df["key"].astype("int32")

    print(union_hole_id_table_df)

    # Rebuild the campaign hole ID category data from the union lookup table and use it for the object.
    union_hole_id_table_component = LookupTable_V1_0_1.from_dict(data_client.save_dataframe(union_hole_id_table_df))

    union_hole_id_values_df = pd.DataFrame()
    union_hole_id_values_df["data"] = [i for i in range(0, len(union_hole_id_table_df))]
    union_hole_id_values_component = IntegerArray1_V1_0_1.from_dict(data_client.save_dataframe(union_hole_id_values_df))

    object_hole_id_component = CategoryData_V1_0_1(
        table=union_hole_id_table_component,
        values=union_hole_id_values_component,
    )
```

### Interim collar coordinates and distances

The interim collar table stores the surface location (`x`, `y`, `z`) and the depth information (`final`, `target`, `current`) for each interim hole. We also extend the campaign bounding box so that it covers the interim collars.


```python
from evo_schemas.components import BoundingBox_V1_0_1
from evo_schemas.elements import FloatArray3_V1_0_1

if has_interim:
    # Collar coordinates (one row per interim hole).
    interim_coordinates_df = interim_collar_df[["x", "y", "z"]].astype(float).reset_index(drop=True)
    interim_coordinates_component = FloatArray3_V1_0_1.from_dict(data_client.save_dataframe(interim_coordinates_df))

    # Collar distances: final, target and current depth (one row per interim hole).
    interim_distances_df = interim_collar_df[["final", "target", "current"]].astype(float).reset_index(drop=True)
    interim_distances_component = FloatArray3_V1_0_1.from_dict(data_client.save_dataframe(interim_distances_df))

    # Extend the campaign bounding box to include the interim collar locations.
    bounding_box = BoundingBox_V1_0_1(
        min_x=min(bounding_box.min_x, interim_coordinates_df["x"].min()),
        max_x=max(bounding_box.max_x, interim_coordinates_df["x"].max()),
        min_y=min(bounding_box.min_y, interim_coordinates_df["y"].min()),
        max_y=max(bounding_box.max_y, interim_coordinates_df["y"].max()),
        min_z=min(bounding_box.min_z, interim_coordinates_df["z"].min()),
        max_z=max(bounding_box.max_z, interim_coordinates_df["z"].max()),
    )
```

### Interim collar attributes

Any columns in the interim collar file that are not part of the fixed schema are published as collar attributes. Each attribute has one value per interim hole.


```python
if has_interim:
    # Publish the extra collar columns as attributes (one value per interim hole).
    interim_collar_reserved_columns = ["hole_id", "x", "y", "z", "final", "target", "current"]
    interim_collar_attributes = create_attributes(interim_collar_df, interim_collar_reserved_columns, data_client)
```

### Interim hole chunks and collars

The hole chunks map each interim hole to the range of survey rows that belong to it. The hole index references the union hole ID lookup table, so interim-only holes are mapped correctly. The chunks, coordinates, distances and attributes are then combined into the interim collar component.


```python
from evo_schemas.components import HoleChunks_V1_0_0, HoleCollars_V1_0_0

if has_interim:
    # Map each interim hole to its range of survey rows.
    interim_hole_chunks_df = create_hole_chunks_df(
        collar_hole_ids=interim_collar_df[interim_collar_hole_id_column],
        path_hole_ids=interim_survey_df[interim_survey_hole_id_column],
        hole_id_table=union_hole_id_table_df,
    )
    print(interim_hole_chunks_df)
    interim_hole_chunks_component = HoleChunks_V1_0_0.from_dict(data_client.save_dataframe(interim_hole_chunks_df))

    interim_hole_collars_component = HoleCollars_V1_0_0(
        coordinates=interim_coordinates_component,
        distances=interim_distances_component,
        holes=interim_hole_chunks_component,
        attributes=interim_collar_attributes,
    )
```

### Interim survey path and section

The interim path is a downhole direction vector with the columns `distance`, `azimuth` and `dip`. Any extra columns in the survey file are published as survey attributes, with one value per survey. The interim collar and path components are then combined into the interim section.


```python
from evo_schemas.components import DownholeDirectionVector_V1_0_0
from evo_schemas.objects import DrillingCampaign_V1_0_0_Interim

if has_interim:
    # The interim path columns are distance, azimuth and dip.
    interim_path_df = interim_survey_df[["depth", "azimuth", "dip"]].rename(columns={"depth": "distance"})
    interim_path_df = interim_path_df.astype(float).reset_index(drop=True)
    interim_path_component = DownholeDirectionVector_V1_0_0.from_dict(data_client.save_dataframe(interim_path_df))

    # Publish the extra survey columns as attributes (one value per survey).
    interim_survey_reserved_columns = ["hole_id", "depth", "dip", "azimuth"]
    interim_path_component.attributes = create_attributes(
        interim_survey_df, interim_survey_reserved_columns, data_client
    )

    # Assemble the interim section from the collar and survey path.
    interim_component = DrillingCampaign_V1_0_0_Interim(
        collar=interim_hole_collars_component,
        path=interim_path_component,
    )
```

### Assemble the drilling campaign object and publish to Evo

Finally, assemble the complete drilling campaign object and publish it to Evo in a single step. The interim section and the union hole ID component are included only when interim data was found; otherwise the object contains just the planned section.


```python
from evo_schemas.objects import DrillingCampaign_V1_0_0

from evo.notebooks import FeedbackWidget

# Whether we are updating an existing object depends on the selection made above.
# object_selector.value is None when "Create a new object" is selected, otherwise it is the
# metadata of the existing object to update.
selected_object = object_selector.value

# When updating, reuse the existing object's UUID and name so it is overwritten in place;
# when creating, use a None UUID (Evo assigns one) and the name defined in the metadata cell.
object_uuid = None if selected_object is None else selected_object.id
target_name = object_name if selected_object is None else selected_object.name

# Assemble the complete geoscience object.
# - The hole ID component is the union of planned and interim IDs when interim data is present,
#   otherwise it is the planned-only component.
# - The interim section is included only when interim data was found (None otherwise).
# - The bounding box already covers the interim collars if they were loaded.
dc = DrillingCampaign_V1_0_0(
    name=target_name,
    bounding_box=bounding_box,
    coordinate_reference_system=coordinate_reference_system,
    uuid=object_uuid,
    hole_id=object_hole_id_component,
    planned=planned_component,
    interim=interim_component,
    tags=object_tags,
)

await data_client.upload_referenced_data(dc.as_dict(), FeedbackWidget("Uploading data"))

if selected_object is None:
    # Create a brand-new object at the configured path.
    new_drilling_campaign_metadata = await object_client.create_geoscience_object(full_obj_path, dc.as_dict())
    action = "Created new"
else:
    # Update the selected object in place, creating a new version of it.
    new_drilling_campaign_metadata = await object_client.update_geoscience_object(dc.as_dict())
    action = "Updated existing"

data_summary = "planned and interim data" if has_interim else "planned data only"
print(f"{action} drilling campaign object {target_name} with {data_summary}!")
```
