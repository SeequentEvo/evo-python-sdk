## Download a drilling campaign object and save it in CSV format

This example shows how to download a drilling-campaign object from an Evo workspace and rebuild the source CSV files from the published data.

The notebook reconstructs the same files that the `create-a-drilling-campaign` notebook consumes:

- `planned.csv` — the planned drillholes.
- `interim-collar.csv` — the interim drillhole collars (`hole_id`, `x`, `y`, `z`, `final`, `target`, `current`).
- `interim-survey.csv` — the interim downhole surveys (`hole_id`, `depth`, `dip`, `azimuth`).

Any collar or survey columns that were published as attributes are restored as extra columns. The reconstructed files can be re-uploaded with the `create-a-drilling-campaign` notebook without any changes to the published data.

> **Note:** This notebook requires an existing drilling campaign object in your Evo workspace. 
>
> If you don't have one yet, run the `create-a-drilling-campaign/sdk-examples.ipynb` notebook first to create a drilling campaign object that you can then download with this notebook.

### Requirements

You must have a Seequent account with the Evo entitlement to use this notebook.

The following parameters must be provided:

- The client ID of your Evo application.
- The callback/redirect URL of your Evo application.

To obtain these app credentials, refer to the [Apps and tokens guide](https://developer.seequent.com/docs/guides/getting-started/apps-and-tokens) in the Seequent Developer Portal.


```python
from evo.colormaps import ColormapAPIClient
from evo.notebooks import FeedbackWidget, ServiceManagerWidget
from evo.objects import ObjectAPIClient

# Evo app credentials
client_id = "<your-client-id>"  # Replace with your client ID
redirect_url = "<your-redirect-url>"  # Replace with your redirect URL

manager = await ServiceManagerWidget.with_auth_code(
    client_id=client_id,
    redirect_url=redirect_url,
).login()
```

### Use the Evo Python SDK to create an object client and a data client

```python
# The object client will manage your auth token and Geoscience Object API requests.
object_client = ObjectAPIClient(manager.get_environment(), manager.get_connector())

# The colormap client will manage your colormap API requests.
colormap_client = ColormapAPIClient(manager.get_environment(), manager.get_connector())

# The data client will manage saving your data as Parquet and publishing your data to Evo storage.
data_client = object_client.get_data_client(manager.cache)
```

### Display all drilling campaigns in the workspace

```python
import ipywidgets as widgets
from IPython.display import display

all_objects = await object_client.list_all_objects()

# Filter drilling campaigns
drilling_campaigns = [obj for obj in all_objects if "drilling-campaign" in obj.schema_id.sub_classification]

if len(drilling_campaigns) == 0:
    print("No drilling campaigns found.")
else:
    # Create dropdown options with name as label and ID as value
    dropdown_options = [(obj.name.removesuffix(".json"), str(obj.id)) for obj in drilling_campaigns]

    drilling_campaign_dropdown = widgets.Dropdown(
        options=dropdown_options,
        description="Drilling campaign:",
        disabled=False,
        style={"description_width": "initial"},
        layout=widgets.Layout(width="600px"),
    )

    print(f"Found {len(drilling_campaigns)} drilling campaign(s). Please select one:")
    display(drilling_campaign_dropdown)
```

Select the drilling campaign object from the dropdown above. 

The cell below downloads the object metadata and prints the number of holes in the drilling campaign.

```python
import json
from datetime import datetime
from uuid import UUID

from pygments import highlight
from pygments.formatters import TerminalTrueColorFormatter
from pygments.lexers import JsonLexer


def serialize_for_display(obj):
    """Convert an object's attributes to a JSON-serializable dictionary."""
    if isinstance(obj, UUID):
        return str(obj)
    elif isinstance(obj, datetime):
        return obj.isoformat()
    elif isinstance(obj, list):
        return [serialize_for_display(item) for item in obj]
    elif isinstance(obj, dict):
        return {key: serialize_for_display(value) for key, value in obj.items()}
    elif hasattr(obj, "__dict__"):
        return {key: serialize_for_display(value) for key, value in vars(obj).items()}
    else:
        return obj


object_id = drilling_campaign_dropdown.value

downloaded_object = await object_client.download_object_by_id(object_id=object_id, version=None)

metadata = downloaded_object.metadata
downloaded_dict = downloaded_object.as_dict()

print("\nDrilling Campaign Structure:\n")
serialized = serialize_for_display(downloaded_dict)
print(highlight(json.dumps(serialized, indent=4), JsonLexer(), TerminalTrueColorFormatter(style="lightbulb")))

number_of_holes = downloaded_dict["hole_id"]["table"]["length"]
print(f"\nThe drilling campaign contains {number_of_holes} holes.")
```

### Reconstruct the source CSV files

The cells below download each table from the published object and rebuild the source CSV files. The helper functions download a table as a dataframe and restore any published attributes as named columns (mapping category attributes back to their text values).

We also download the campaign hole ID lookup table, which is the union of the planned and interim hole IDs, and use it to map each hole index back to its name.


```python
import os

import pandas as pd

# Directory where the reconstructed CSV files will be written. These files match the input format
# expected by the create-a-drilling-campaign notebook, so they can be re-uploaded without any
# changes to the published data.
output_path = "downloaded-data"
os.makedirs(output_path, exist_ok=True)


def download_table(table_info, description="table"):
    """Download a table by its table info. Returns a coroutine that resolves to a pyarrow table."""
    return data_client.download_table(
        object_id=metadata.id,
        version_id=None,
        table_info=table_info,
        fb=FeedbackWidget(f"Downloading {description} as '{table_info['data']}'"),
    )


async def download_attribute(attribute):
    """Download a single attribute and return it as a named pandas Series (one value per row)."""
    name = attribute["name"]
    attribute_type = attribute["attribute_type"]

    values = (await download_table(attribute["values"], f"attribute '{name}'")).to_pandas()
    values.columns = ["data"]

    if attribute_type == "category":
        # Map the integer keys back to their text values using the attribute's lookup table.
        lookup = (await download_table(attribute["table"], f"attribute '{name}' lookup table")).to_pandas()
        merged = values.merge(lookup, left_on="data", right_on="key", how="left")
        return merged["value"].rename(name).reset_index(drop=True)

    return values["data"].rename(name).reset_index(drop=True)


# Download the campaign hole ID lookup table (maps each hole key to its name). This is the union of
# the planned and interim hole IDs.
hole_id_table = (await download_table(downloaded_dict["hole_id"]["table"], "hole ID lookup table")).to_pandas()
hole_index_to_name = dict(zip(hole_id_table["key"], hole_id_table["value"]))
```

```python
# Reconstruct planned.csv from the published planned section.
planned = downloaded_dict["planned"]

planned_coordinates = (await download_table(planned["collar"]["coordinates"], "planned collar coordinates")).to_pandas()
planned_distances = (await download_table(planned["collar"]["distances"], "planned collar distances")).to_pandas()
planned_chunks = (await download_table(planned["collar"]["holes"], "planned collar hole chunks")).to_pandas()
planned_path = (await download_table(planned["path"], "planned path")).to_pandas()

# Align each collar to its path row using the hole chunks (planned holes have one path row each).
planned_path_rows = pd.concat(
    [
        planned_path.iloc[int(offset) : int(offset) + int(count)].reset_index(drop=True)
        for offset, count in zip(planned_chunks["offset"], planned_chunks["count"])
    ],
    axis=0,
).reset_index(drop=True)

planned_df = pd.DataFrame()
planned_df["Holeid"] = [hole_index_to_name[hole_index] for hole_index in planned_chunks["hole_index"]]
planned_df["Easting"] = planned_coordinates["x"]
planned_df["Northing"] = planned_coordinates["y"]
planned_df["Elevation"] = planned_coordinates["z"]
planned_df["Azimuth"] = planned_path_rows["azimuth"]
planned_df["Dip"] = planned_path_rows["dip"]
planned_df["Lift Rate"] = planned_path_rows["lift rate"]
planned_df["Drift Rate"] = planned_path_rows["drift rate"]
planned_df["Distance"] = planned_path_rows["deviation rate distance"]
# The create notebook derives final = target + extension, so extension = final - target.
planned_df["Extension"] = planned_distances["final"] - planned_distances["target"]
planned_df["Target Depth"] = planned_distances["target"]

# Restore any planned collar attributes as extra columns.
for attribute in planned["collar"].get("attributes") or []:
    planned_df[attribute["name"]] = await download_attribute(attribute)

planned_file = os.path.join(output_path, "planned.csv")
planned_df.to_csv(planned_file, index=False)
print(f"✓ Saved {len(planned_df)} planned holes to '{planned_file}'")
planned_df.head()
```

### Reconstruct the interim files

If the drilling campaign has an interim section, rebuild `interim-collar.csv` and `interim-survey.csv`.

- The interim collar table provides the hole locations (`x`, `y`, `z`) and depths (`final`, `target`, `current`), with one row per interim hole.
- The interim survey table provides the downhole path (`depth`, `dip`, `azimuth`), with one row per survey. Each survey's `hole_id` is recovered from the collar hole chunks.

Any extra collar or survey columns are restored as attributes.


```python
from IPython.display import display

# Reconstruct the interim files from the published interim section (if present).
if downloaded_dict.get("interim"):
    interim = downloaded_dict["interim"]

    # ---- Interim collar (one row per interim hole) ----
    interim_coordinates = (
        await download_table(interim["collar"]["coordinates"], "interim collar coordinates")
    ).to_pandas()
    interim_distances = (await download_table(interim["collar"]["distances"], "interim collar distances")).to_pandas()
    interim_chunks = (await download_table(interim["collar"]["holes"], "interim collar hole chunks")).to_pandas()

    interim_collar_df = pd.DataFrame()
    interim_collar_df["hole_id"] = [hole_index_to_name[hole_index] for hole_index in interim_chunks["hole_index"]]
    interim_collar_df["x"] = interim_coordinates["x"]
    interim_collar_df["y"] = interim_coordinates["y"]
    interim_collar_df["z"] = interim_coordinates["z"]
    interim_collar_df["final"] = interim_distances["final"]
    interim_collar_df["target"] = interim_distances["target"]
    interim_collar_df["current"] = interim_distances["current"]

    for attribute in interim["collar"].get("attributes") or []:
        interim_collar_df[attribute["name"]] = await download_attribute(attribute)

    interim_collar_file = os.path.join(output_path, "interim-collar.csv")
    interim_collar_df.to_csv(interim_collar_file, index=False)
    print(f"✓ Saved {len(interim_collar_df)} interim holes to '{interim_collar_file}'")

    # ---- Interim survey (one row per survey) ----
    interim_path = (await download_table(interim["path"], "interim survey path")).to_pandas()

    # Recover each survey row's hole ID from the collar hole chunks.
    survey_hole_ids = [None] * len(interim_path)
    for hole_index, offset, count in zip(
        interim_chunks["hole_index"], interim_chunks["offset"], interim_chunks["count"]
    ):
        name = hole_index_to_name[hole_index]
        for row in range(int(offset), int(offset) + int(count)):
            survey_hole_ids[row] = name

    interim_survey_df = pd.DataFrame()
    interim_survey_df["hole_id"] = survey_hole_ids
    interim_survey_df["depth"] = interim_path["distance"]
    interim_survey_df["dip"] = interim_path["dip"]
    interim_survey_df["azimuth"] = interim_path["azimuth"]

    for attribute in interim["path"].get("attributes") or []:
        interim_survey_df[attribute["name"]] = await download_attribute(attribute)

    interim_survey_file = os.path.join(output_path, "interim-survey.csv")
    interim_survey_df.to_csv(interim_survey_file, index=False)
    print(f"✓ Saved {len(interim_survey_df)} interim surveys to '{interim_survey_file}'")

    display(interim_collar_df.head())
    display(interim_survey_df.head())
else:
    print("This drilling campaign has no interim section; only planned.csv was written.")
```
